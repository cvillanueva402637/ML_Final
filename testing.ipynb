```python
import joblib
import pandas as pd

# Load model and feature names
model = joblib.load('student_exam_rf_model.pkl')
features = joblib.load('model_features.pkl')

# Example: new student data (must match features)
new_data = pd.DataFrame([{
    'Hours_Studied': 5,
    'Previous_Score': 70,
    'Sleep_Hours': 7,
    'Extracurricular_Activities_Yes': 1
    # Add other columns as needed
}])

# Align with training columns (handle missing ones)
for col in features:
    if col not in new_data.columns:
        new_data[col] = 0
new_data = new_data[features]

# Predict
predicted_score = model.predict(new_data)
print("Predicted Exam Score:", predicted_score[0])
```
